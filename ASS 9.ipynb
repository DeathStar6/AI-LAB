```python
def evaluate_student(attendance, marks, assignment, internal):
    """
    Evaluates student performance based on rule-based logic
    """

    # Rule 1: Excellent
    if (attendance >= 85 and marks >= 80 and 
        assignment >= 80 and internal >= 80):
        return "Excellent"

    # Rule 2: Good
    elif (attendance >= 75 and marks >= 65 and 
          assignment >= 65 and internal >= 65):
        return "Good"

    # Rule 3: Average
    elif (attendance >= 60 and marks >= 50 and 
          assignment >= 50 and internal >= 50):
        return "Average"

    # Rule 4: Poor
    else:
        return "Poor"


# Taking user input
attendance = float(input("Enter Attendance (%): "))
marks = float(input("Enter Marks (%): "))
assignment = float(input("Enter Assignment Score (%): "))
internal = float(input("Enter Internal Assessment (%): "))

# Evaluate performance
result = evaluate_student(attendance, marks, assignment, internal)

# Output result
print("\nStudent Performance:", result)
```
