```python
def minimax(depth, index, is_max, values):
    if depth == 3:  
        return values[index]

    if is_max:
        return max(
            minimax(depth + 1, index * 2, False, values),
            minimax(depth + 1, index * 2 + 1, False, values)
        )
    else:
        return min(
            minimax(depth + 1, index * 2, True, values),
            minimax(depth + 1, index * 2 + 1, True, values)
        )


values = list(map(int, input("Enter 8 leaf node values (space-separated): ").split()))

if len(values) != 8:
    print("Please enter exactly 8 values.")
else:
    optimal_score = minimax(0, 0, True, values)
    print("Optimal value for MAX player is:", optimal_score)
```

```python


def alphabeta(depth, index, is_max, values, alpha, beta):
    # Leaf node
    if depth == 3:   # 8 leaf nodes → height = 3
        return values[index]

    if is_max:
        best = float('-inf')

        # Left child
        val = alphabeta(depth + 1, index * 2, False, values, alpha, beta)
        best = max(best, val)
        alpha = max(alpha, best)

        # Prune check
        if beta <= alpha:
            print(f"Pruned branch at depth {depth}, index {index*2 + 1}")
            return best

        # Right child
        val = alphabeta(depth + 1, index * 2 + 1, False, values, alpha, beta)
        best = max(best, val)
        alpha = max(alpha, best)

        return best

    else:
        best = float('inf')

        # Left child
        val = alphabeta(depth + 1, index * 2, True, values, alpha, beta)
        best = min(best, val)
        beta = min(beta, best)

        # Prune check
        if beta <= alpha:
            print(f"Pruned branch at depth {depth}, index {index*2 + 1}")
            return best

        # Right child
        val = alphabeta(depth + 1, index * 2 + 1, True, values, alpha, beta)
        best = min(best, val)
        beta = min(beta, best)

        return best


# Given leaf nodes
values = [3, 5, 2, 9, 12, 5, 23, 23]

# Run Alpha-Beta
optimal_score = alphabeta(0, 0, True, values, float('-inf'), float('inf'))

print("\nOptimal value for MAX player is:", optimal_score)
```
